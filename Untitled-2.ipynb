```python
from lmxy import OpenAiLike, get_tokenizer

tokenizer = get_tokenizer('Qwen/Qwen3-VL-30B-A3B')
# tokenizer = get_tokenizer('Qwen/Qwen3.5-35B-A3B')
tokenizer = get_tokenizer('Qwen/Qwen3.6-35B-A3B')
llm = OpenAiLike(
    model='qwen3-vl-30b-a3b',
    # model='qwen3.5-35b-a3b',
    # model='qwen3.6-35b-a3b',
    base_url='http://172.16.216.100:11434/v1',
    timeout=None,
    context_window=1000,
)
```

```python
from llama_index.core.schema import NodeWithScore, TextNode

with open(r'D:/Downloads/промт.txt') as f:
    nodes = [NodeWithScore(node=TextNode(text=f.read()))]

print(len(tokenizer(nodes[0].get_content())))
```

```python
from lmxy.synth import synthesize

tks = await synthesize(
    *nodes,
    query='привет',
    llm=llm,
    tokenizer=tokenizer,
    padding=50,
)
print('>>>')
async for tk in tks:
    print(tk, end='')
```
